```python
from sympy import symbols, exp, sin, diff
x = symbols('x')
f_exp = x ** 2
print(f_exp)

g_expr = exp(-2 * x) + (3 * sin(3 * x))
print(g_expr)

df_expr = diff(f_exp)
dg_expr = diff(g_expr)

print(df_expr, dg_expr)
```

```python
import numpy as np
from sympy import lambdify
print(dg_expr.evalf(subs={x:1}))
dg_func = lambdify(x,dg_expr,'numpy')
a = np.array([1,2,3])
dg_func(a)


```

```python
func = (abs(x))
der_func = diff(func)
func_1 = lambdify(x,der_func,'numpy')
func_1(10)
```

```python
xr = symbols('x', real=True)
func_r = abs(xr)
der_r = diff(func_r)
print(der_r)
func_2 = lambdify(xr, der_r, 'numpy')
func_2(np.array([-2., 0., 2.]))
```

** Numpy Based Derivatives**

```python
import numpy as np

x = np.linspace(-5, 5, 100)
x_square = x ** 2
slopes = np.gradient(x_square,x)
#print(slopes)

difference = np.abs(slopes - 2*x)
np.max(difference)
```

```python
x_abs = np.abs(x)
#print(x_abs)
slopes = np.gradient(x_abs)
slopes[47:53]
```

```python
import jax.numpy as jnp
from jax import grad, vmap


def function (x):
    return x**2

df = grad(function)
print(df(3.0))
```

```python
a = jnp.array([1.0,2.0,3.0])
df_many = vmap(df)
print(df_many)
print(df_many(a))
```
